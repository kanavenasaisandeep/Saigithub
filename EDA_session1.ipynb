**Task-2**
- seperate categorical columns and numerical columns

```python
keys = visa_df.dtypes.keys() 
values = visa_df.dtypes.values
```

```python
categorical = [] 
numerical = [] 
for i,j in zip(keys,values):    
    if j == 'object':        
        categorical.append(i)    
    else:        
        numerical.append(i)
```

```python
print(categorical) 
print(numerical)
```

```python
data = [] 
pd.DataFrame(zip(categorical,numerical),columns = ['categorical','numerical'])
```
